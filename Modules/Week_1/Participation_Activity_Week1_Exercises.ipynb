## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
mean=sum(salaries)/len(salaries)

import statistics
statistics.mean(salaries)

print(mean)
print(statistics.mean(salaries))
print((mean)==(statistics.mean(salaries)))
```

### median

```python
sorted_salaries=sorted(salaries)
median=(sorted_salaries[49]+sorted_salaries[50])/2

statistics.median(salaries)

print(median)
print(statistics.median(salaries))
print((median)==(statistics.median(salaries)))
```

### mode

```python
mode_salaries=max(salaries,key=salaries.count)

statistics.mode(salaries)

print(mode_salaries)
print(statistics.mode(salaries))
print((mode_salaries)==(statistics.mode(salaries)))
```

### sample variance
Remember to use Bessel's correction.

```python
variance=sum((x-mean)**2 for x in salaries)/(len(salaries)-1)

statistics.variance(salaries)

print(variance)
print(statistics.variance(salaries))
print((variance)==(statistics.variance(salaries)))
```

### sample standard deviation
Remember to use Bessel's correction.

```python
standard_deviation=variance**0.5

print(standard_deviation)
print(statistics.stdev(salaries))
print((standard_deviation)==(statistics.stdev(salaries)))
```

## Exercise 6: Calculating more statistics
### range

```python
range_salaries=max(salaries)-min(salaries)
range_salaries
```

### coefficient of variation

```python
cofficient_of_variation=standard_deviation/mean
cofficient_of_variation
```

### interquartile range

```python
lower_half=sorted_salaries[:50]
upper_half=sorted_salaries[50:]

Q1=(lower_half[24]+lower_half[25])/2
Q3=(upper_half[24]+upper_half[25])/2
IQR=Q3-Q1
IQR
```

### quartile coefficent of dispersion

```python
quartile_coefficent=IQR/(Q1+Q3)
quartile_coefficent
```
