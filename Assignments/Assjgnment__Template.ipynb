```python
import os
import subprocess

owner = 'EstelleYang111'
repository = 'MSSP607'

repo_path = f'/content/{repository}'

# Clone the public GitHub repository (no token required)
if not os.path.exists(repo_path):
    subprocess.run(
        ['git', 'clone',
         f'https://github.com/{owner}/{repository}.git',
         repo_path],
        check=True
    )

# Navigate into the cloned repository directory
os.chdir(repo_path)

print(f"Changed directory to: {os.getcwd()}")
```
