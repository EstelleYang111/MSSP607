```python
import pandas as pd

url = "https://raw.githubusercontent.com/EstelleYang111/MSSP607/main/data/week12/Student%20Performance%20Data%20%281%29.csv"

df = pd.read_csv(url)

print(df.shape)
display(df.head())
print(df.columns.tolist())
```

```python
male_completed = df[(df["gender"] == "male") & (df["test preparation course"] == "completed")]["total score"]

male_none = df[(df["gender"] == "male") & (df["test preparation course"] == "none")]["total score"]

print("Male completed:", len(male_completed))
print("Male none:", len(male_none))

print("Mean score (completed):", male_completed.mean())
print("Mean score (none):", male_none.mean())
```

```python
from scipy import stats

t_stat, p_value = stats.ttest_ind(male_completed, male_none, equal_var=False)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

Yes. An independent-samples Welch's t-test found a statistically significant difference in total scores between male students who completed the test preparation course and those who did not (t = 6.18, p < 0.001). Male students who completed the course had a higher average score (212.34) than those who did not (189.13).

```python
female_completed = df[(df["gender"] == "female") & (df["test preparation course"] == "completed")]["total score"]
female_none = df[(df["gender"] == "female") & (df["test preparation course"] == "none")]["total score"]

t_stat_f, p_value_f = stats.ttest_ind(female_completed, female_none, equal_var=False)

print("Female completed:", len(female_completed))
print("Female none:", len(female_none))
print("Mean score (completed):", female_completed.mean())
print("Mean score (none):", female_none.mean())
print("T-statistic:", t_stat_f)
print("P-value:", p_value_f)
```

Another interesting finding is that female students who completed the test preparation course also had significantly higher total scores than those who did not. The average score was 223.36 for female students who completed the course, compared with 200.63 for those who did not. A Welch's t-test confirmed that this difference was statistically significant (t = 6.13, p < 0.001).
