```python
try:
  number = int(input("Enter a number: "))
  print(number)
except ValueError:
  print("Please enter a valid number.")

# A program can use try and except statements to handle errors and prevent the program from crashing.
```

```python
def collatz(number):
  # If the number is even, divide it by 2.
  if number % 2 == 0:
    result = number // 2
  else:
    # If the number is odd, multiply it by 3 and add 1.
    result = 3 * number + 1

  print(result)
  return result

try:
  number = int(input("Enter a number: "))

  # Continue the sequence until the number reaches 1.
  while number != 1:
    number = collatz(number)

except ValueError:
    print("Please enter a valid integer.")
```
