```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/MSSP607/Modules/Week06/Student Performance Data.csv')

df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
high_performers = lambda data: data.loc[
    data['total score'] >= 240,
    ['gender', 'math score', 'reading score',
     'writing score', 'total score', 'admission prospect']
].copy()
```

```python
high_score_df = high_performers(df)

high_score_df.head()
```

```python
print("Original DataFrame shape:", df.shape)
print("New DataFrame shape:", high_score_df.shape)
print("Minimum total score:", high_score_df['total score'].min())
```

I used a lambda function to create a new DataFrame from the original student performance dataset. The function selects students with a total score of at least 240 and keeps six columns related to their academic performance and admission prospects. This allows me to focus on a smaller group of students while keeping the original dataset unchanged.
