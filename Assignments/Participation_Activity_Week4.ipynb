```python
ring_of_fire = [
    'Bolivia', 'Chile', 'Ecuador', 'Peru', 'Costa Rica',
    'Guatemala', 'Mexico', 'Japan', 'Philippines', 'Indonesia',
    'New Zealand', 'Antarctic', 'Canada', 'Fiji', 'Alaska',
    'Washington', 'CA', 'Russia', 'Taiwan', 'Tonga',
    'Kermadec Islands'
]
```

```python
pattern = '|'.join(ring_of_fire)
pattern
```

```python
df['ring_of_fire'] = df['place'].str.contains(pattern, na=False)

df.loc[df['place'].str.contains('New Mexico', na=False), 'ring_of_fire'] = False

df[['place', 'ring_of_fire']].head()
```

```python
df['ring_of_fire'].value_counts()
```

```python
ring_earthquakes = df[df['ring_of_fire'] == True]
ring_earthquakes['tsunami'].value_counts()
```
