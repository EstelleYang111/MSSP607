```python
letters=['a','b','c','d']
print(letters[int(int('3'*2)//11)])
# '3' * 2 becomes '33', int('33') becomes 33,
# 33 // 11 equals 3, so letters[3] is 'd'.

print(letters[-1])
# Index -1 refers to the last item in the list, so it returns 'd'.

print(letters[:2])
# [:2] returns the items before index 2, so it returns 'a' and 'b'.
```

```python
spam = ['apples', 'bananas', 'tofu', 'cats']
def comma_code(items):
   return ",".join(items[:-1]) + ', and ' + items[-1]
print(comma_code(spam))
```
