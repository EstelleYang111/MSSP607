```python
base_url = "https://raw.githubusercontent.com/EstelleYang111/MSSP607/main/data/Final_Project/"

cf = pd.read_csv(
    base_url + "Cystic%20Fibrosis-studies.csv",
    keep_default_na=False,
    na_values=[""]
)

hd = pd.read_csv(
    base_url + "Huntington%20Disease-studies.csv",
    keep_default_na=False,
    na_values=[""]
)

dmd = pd.read_csv(
    base_url + "Duchenne%20Muscular%20Dystrophy-studies.csv",
    keep_default_na=False,
    na_values=[""]
)

cf["Disease"] = "Cystic Fibrosis"
hd["Disease"] = "Huntington Disease"
dmd["Disease"] = "Duchenne Muscular Dystrophy"

combined_df = pd.concat(
    [cf, hd, dmd],
    ignore_index=True
)

print("=== Combined Dataset ===")
print("Shape:", combined_df.shape)

print("\n=== Studies by Disease ===")
display(combined_df["Disease"].value_counts())

print("\n=== Duplicate NCT Numbers ===")
print(combined_df["NCT Number"].duplicated().sum())

display(combined_df[["NCT Number", "Disease", "Study Status", "Start Date"]].head())
```

```python
import matplotlib.pyplot as plt

trial_counts = (
    combined_df.groupby("Disease")["NCT Number"]
    .nunique()
    .sort_values(ascending=False)
)

plt.figure(figsize=(11, 6))

bars = plt.bar(
    trial_counts.index,
    trial_counts.values,
    color=["#2563A6", "#58A6A6", "#9B8BC4"],
    width=0.6
)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 8,
        str(int(height)),
        ha="center",
        fontsize=11
    )

plt.title(
    "Clinical Trial Volume Across Three Rare Diseases (2016–2025)",
    fontsize=14,
    fontweight="bold",
    pad=18
)

plt.xlabel("Rare Disease", fontsize=12)
plt.ylabel("Number of Clinical Trials", fontsize=12)

plt.ylim(0, 650)
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.gca().set_axisbelow(True)

plt.tight_layout()

plt.savefig(
    "research_volume_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
status_counts = pd.crosstab(
    combined_df["Disease"],
    combined_df["Study Status"]
)

status_percentages = (
    status_counts.div(
        status_counts.sum(axis=1),
        axis=0
    ) * 100
).round(2)

print("=== Clinical Trial Status Counts ===")
display(status_counts)

print("\n=== Clinical Trial Status Percentages ===")
display(status_percentages)

print("\n=== Percentage Check ===")
display(status_percentages.sum(axis=1))
```

```python
import matplotlib.pyplot as plt

status_pct = status_counts.div(
    status_counts.sum(axis=1),
    axis=0
) * 100

status_order = [
    "COMPLETED",
    "RECRUITING",
    "ACTIVE_NOT_RECRUITING",
    "TERMINATED",
    "WITHDRAWN",
    "UNKNOWN",
    "NOT_YET_RECRUITING",
    "ENROLLING_BY_INVITATION",
    "SUSPENDED"
]

status_colors = [
    "#2563A6",
    "#58A6A6",
    "#9B8BC4",
    "#D98C62",
    "#C85A5A",
    "#A0A0A0",
    "#E4BC62",
    "#7E9E64",
    "#725D86"
]

fig, ax = plt.subplots(figsize=(12, 6))

left = [0] * len(status_pct)

for status, color in zip(status_order, status_colors):
    values = status_pct[status]

    ax.barh(
        status_pct.index,
        values,
        left=left,
        label=status.replace("_", " ").title(),
        color=color,
        edgecolor="white",
        linewidth=0.5
    )


    left = [
        l + v for l, v in zip(left, values)
    ]

ax.set_title(
    "Clinical Trial Status Distribution Across Three Rare Diseases",
    fontsize=14,
    fontweight="bold",
    pad=18
)

ax.set_xlabel("Percentage of Clinical Trials (%)", fontsize=12)
ax.set_ylabel("Rare Disease", fontsize=12)

ax.set_xlim(0, 100)
ax.invert_yaxis()
ax.grid(axis="x", linestyle="--", alpha=0.2)
ax.set_axisbelow(True)

ax.legend(
    title="Study Status",
    loc="upper center",
    bbox_to_anchor=(0.5, -0.15),
    ncol=3,
    fontsize=9
)

plt.tight_layout()

plt.savefig(
    "clinical_trial_status_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
us_sites = []

for _, row in combined_df.iterrows():


    if pd.isna(row["Locations"]):
        continue


    locations = row["Locations"].split("|")

    for location in locations:


        parts = [part.strip() for part in location.split(",")]


        if len(parts) < 3 or parts[-1] != "United States":
            continue


        if any(char.isdigit() for char in parts[-2]):
            state = parts[-3]
        else:
            state = parts[-2]


        us_sites.append({
            "NCT Number": row["NCT Number"],
            "Disease": row["Disease"],
            "State": state,
            "Location": location
        })

us_sites_df = pd.DataFrame(us_sites)

print("=== US Site Records ===")
print("Total site entries:", len(us_sites_df))

print("\n=== US Site Entries by Disease ===")
display(us_sites_df["Disease"].value_counts())

print("\n=== Unique US Studies by Disease ===")
display(
    us_sites_df.groupby("Disease")["NCT Number"].nunique()
)

print("\n=== Top 10 States ===")
display(us_sites_df["State"].value_counts().head(10))

print("\n=== Data Preview ===")
display(us_sites_df.head())
```

```python
valid_states = set("""
Alabama|Alaska|Arizona|Arkansas|California|Colorado|
Connecticut|Delaware|Florida|Georgia|Hawaii|Idaho|
Illinois|Indiana|Iowa|Kansas|Kentucky|Louisiana|
Maine|Maryland|Massachusetts|Michigan|Minnesota|
Mississippi|Missouri|Montana|Nebraska|Nevada|
New Hampshire|New Jersey|New Mexico|New York|
North Carolina|North Dakota|Ohio|Oklahoma|Oregon|
Pennsylvania|Rhode Island|South Carolina|South Dakota|
Tennessee|Texas|Utah|Vermont|Virginia|Washington|
West Virginia|Wisconsin|Wyoming|District of Columbia
""".replace("\n", "").split("|"))

valid_states = {state.strip() for state in valid_states}

invalid_sites = us_sites_df[
    ~us_sites_df["State"].isin(valid_states)
]

print("=== State Validation ===")
print("Total site entries:", len(us_sites_df))
print("Invalid or unrecognized state entries:", len(invalid_sites))

print("\n=== Unrecognized State Names ===")
display(invalid_sites["State"].value_counts())

print("\n=== Duplicate Site Records ===")

duplicate_sites = us_sites_df.duplicated(
    subset=["NCT Number", "Location"]
).sum()

print("Duplicate study-location pairs:", duplicate_sites)

print("\n=== Geographic Coverage ===")

represented_states = us_sites_df.loc[
    us_sites_df["State"].isin(valid_states),
    "State"
].nunique()

print("States and DC represented:", represented_states)
```

```python
state_abbr = {
    "Alabama": "AL", "Alaska": "AK",
    "Arizona": "AZ", "Arkansas": "AR",
    "California": "CA", "Colorado": "CO",
    "Connecticut": "CT", "Delaware": "DE",
    "Florida": "FL", "Georgia": "GA",
    "Hawaii": "HI", "Idaho": "ID",
    "Illinois": "IL", "Indiana": "IN",
    "Iowa": "IA", "Kansas": "KS",
    "Kentucky": "KY", "Louisiana": "LA",
    "Maine": "ME", "Maryland": "MD",
    "Massachusetts": "MA", "Michigan": "MI",
    "Minnesota": "MN", "Mississippi": "MS",
    "Missouri": "MO", "Montana": "MT",
    "Nebraska": "NE", "Nevada": "NV",
    "New Hampshire": "NH", "New Jersey": "NJ",
    "New Mexico": "NM", "New York": "NY",
    "North Carolina": "NC", "North Dakota": "ND",
    "Ohio": "OH", "Oklahoma": "OK",
    "Oregon": "OR", "Pennsylvania": "PA",
    "Rhode Island": "RI", "South Carolina": "SC",
    "South Dakota": "SD", "Tennessee": "TN",
    "Texas": "TX", "Utah": "UT",
    "Vermont": "VT", "Virginia": "VA",
    "Washington": "WA", "West Virginia": "WV",
    "Wisconsin": "WI", "Wyoming": "WY",
    "District of Columbia": "DC"
}

state_counts = (
    us_sites_df.groupby(["Disease", "State"])
    .size()
    .reset_index(name="Site Entries")
)

all_combinations = pd.MultiIndex.from_product(
    [combined_df["Disease"].unique(), state_abbr.keys()],
    names=["Disease", "State"]
).to_frame(index=False)

state_counts = all_combinations.merge(
    state_counts,
    on=["Disease", "State"],
    how="left"
)

state_counts["Site Entries"] = (
    state_counts["Site Entries"].fillna(0).astype(int)
)

state_counts["Percentage"] = (
    state_counts["Site Entries"]
    / state_counts.groupby("Disease")["Site Entries"].transform("sum")
    * 100
)

state_counts["State Code"] = (
    state_counts["State"].map(state_abbr)
)

print("=== State-Level Geographic Data ===")
display(state_counts.head(10))

print("\n=== Total Site Entries by Disease ===")
display(state_counts.groupby("Disease")["Site Entries"].sum())

print("\n=== Percentage Check ===")
display(
    state_counts.groupby("Disease")["Percentage"].sum().round(2)
)

print("\n=== Missing State Codes ===")
print(state_counts["State Code"].isna().sum())
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

disease_order = [
    "Cystic Fibrosis",
    "Duchenne Muscular Dystrophy",
    "Huntington Disease"
]

fig = make_subplots(
    rows=1,
    cols=3,
    specs=[[{"type": "geo"}] * 3],
    subplot_titles=disease_order,
    horizontal_spacing=0.03
)

max_percentage = state_counts["Percentage"].max()

for i, disease in enumerate(disease_order):

    disease_data = state_counts[
        state_counts["Disease"] == disease
    ]

    fig.add_trace(
        go.Choropleth(
            locations=disease_data["State Code"],
            z=disease_data["Percentage"],
            locationmode="USA-states",
            customdata=disease_data[
                ["State", "Site Entries"]
            ].to_numpy(),
            coloraxis="coloraxis",
            marker_line_color="white",
            marker_line_width=0.5,
            hovertemplate=(
                "<b>%{customdata[0]}</b><br>"
                "Site Entries: %{customdata[1]}<br>"
                "Share of US Sites: %{z:.2f}%"
                "<extra></extra>"
            )
        ),
        row=1,
        col=i + 1
    )

fig.update_geos(
    scope="usa",
    projection_type="albers usa",
    showlakes=False,
    showcoastlines=False
)

fig.update_layout(
    title=dict(
        text="Geographic Distribution of Clinical Trial Sites Across Three Rare Diseases",
        x=0.5,
        font=dict(size=19)
    ),
    coloraxis=dict(
        colorscale="Blues",
        cmin=0,
        cmax=max_percentage,
        colorbar=dict(
            title="Share (%)",
            len=0.7,
            x=1.01
        )
    ),
    width=1500,
    height=500,
    margin=dict(l=10, r=100, t=100, b=20)
)

fig.show(
    config={
        "displayModeBar": True,
        "toImageButtonOptions": {
            "format": "png",
            "filename": "figure7_us_site_distribution",
            "scale": 2
        }
    }
)

fig.write_html("figure7_us_site_distribution.html")
```
