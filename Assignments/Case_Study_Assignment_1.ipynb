```python
import matplotlib.pyplot as plt
```

```python
gender_mean = df.groupby('GENDER')['G_SC'].mean()
```

```python
gender_mean.plot(kind='bar')
plt.title('Average Global Score by Gender')
plt.ylabel('Average Global Score')
plt.xticks(rotation=0)
for i, value in enumerate(gender_mean):
    plt.text(i, value, f'{value:.2f}', ha='center', va='bottom')
```

```python
stratum_mean = df.groupby('STRATUM')['G_SC'].mean()
```

```python
stratum_plot = stratum_mean.drop(0)
```

```python
stratum_plot.plot(kind='bar')
plt.title('Average Global Score by Stratum')
plt.xlabel('Stratum')
plt.ylabel('Average Global Score')
plt.xticks(rotation=45)

for i, value in enumerate(stratum_plot):
    plt.text(i, value, f'{value:.2f}', ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

```python
father_mean = df.groupby('EDU_FATHER')['G_SC'].mean()
```

```python
father_mean.plot(kind='barh')
plt.title('Average Global Score by Father Education')
plt.xlabel('Average Global Score')
plt.ylabel('Father Education')

for i, value in enumerate(father_mean):
    plt.text(value, i, f'{value:.2f}', va='center')

plt.tight_layout()
plt.show()
```

```python
mother_mean = df.groupby('EDU_MOTHER')['G_SC'].mean()
```

```python
mother_mean.plot(kind='barh')
plt.title('Average Global Score by Mother Education')
plt.xlabel('Average Global Score')
plt.ylabel('Mother Education')

for i, value in enumerate(mother_mean):
    plt.text(value, i, f'{value:.2f}', va='center')

plt.tight_layout()
plt.show()
```
